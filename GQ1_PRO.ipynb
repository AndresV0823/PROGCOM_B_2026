```python
import random

class Objeto:
    def accion(self):
        pass


class Perro(Objeto):
    def accion(self):
        print("🔊 Emite sonidos cortos y repetitivos, suele reaccionar a extraños")


class Gato(Objeto):
    def accion(self):
        print("🔊 Sonido suave, independiente, activo en la noche")


class Robot(Objeto):
    def accion(self):
        print("⚙️ Ejecuta instrucciones de forma lógica y precisa")


class Carro(Objeto):
    def accion(self):
        print("🚘 Se desplaza rápidamente y produce vibraciones constantes")


class Persona(Objeto):
    def accion(self):
        print("🧠 Se comunica, razona y toma decisiones")


class Ave(Objeto):
    def accion(self):
        print("🕊️ Genera sonidos agudos y puede desplazarse por el aire")


class Telefono(Objeto):
    def accion(self):
        print("📱 Emite notificaciones y permite comunicación a distancia")


objetos = [
    ("Perro", Perro()),
    ("Gato", Gato()),
    ("Robot", Robot()),
    ("Carro", Carro()),
    ("Persona", Persona()),
    ("Ave", Ave()),
    ("Telefono", Telefono())
]


def mostrar_opciones(opciones):
    print("\n¿Quién crees que ejecutó accion()? 👀")
    for i, op in enumerate(opciones, start=1):
        print(f"{i}. {op}")


def jugar():
    print("🎮 DESAFÍO DE POLIMORFISMO")
    print("Se ejecuta SIEMPRE el mismo método: accion()")
    print("Pero... ¿quién lo hizo? 🤯\n")

    puntos = 0
    rondas = 5

    seleccion = random.sample(objetos, rondas)

    for i, (nombre_correcto, objeto) in enumerate(seleccion, start=1):
        print("\n========================")
        print(f"Ronda {i}")

        # 🔥 AQUÍ SE HACE MÁS CLARO EL CONCEPTO
        print("\n>>> Ejecutando: objeto.accion() <<<\n")
        objeto.accion()

        nombres = [n for n, _ in objetos]
        opciones = random.sample(nombres, 4)

        if nombre_correcto not in opciones:
            opciones[random.randint(0, 3)] = nombre_correcto

        random.shuffle(opciones)

        mostrar_opciones(opciones)
        respuesta = input("Tu respuesta: ")

        if respuesta.isdigit() and 1 <= int(respuesta) <= len(opciones):
            elegido = opciones[int(respuesta) - 1]

            if elegido == nombre_correcto:
                print("✅ Correcto! +1 punto")
                puntos += 1
            else:
                print(f"❌ Incorrecto. Era: {nombre_correcto}")
        else:
            print(f"❌ Entrada inválida. Era: {nombre_correcto}")

    print("\n========================")
    print(f"🎯 Puntaje final: {puntos}/{rondas}")

    print("\n🧠 CONCLUSIÓN FINAL:")
    print("Durante todo el juego usamos el mismo método: objeto.accion()")
    print("Pero cada objeto hizo algo diferente.")
    print("👉 Eso es POLIMORFISMO 🚀")


jugar()
```
