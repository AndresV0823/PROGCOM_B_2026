```python
import random

# =========================
# 10 PREGUNTAS EXACTAS
# =========================
preguntas = [
    {"p": "¿Qué es el polimorfismo?", "o": ["1. Muchas variables","2. Un método con diferentes comportamientos","3. Error"], "r": "2"},
    {"p": "¿Qué permite el polimorfismo?", "o": ["1. Mismo método en distintas clases","2. Eliminar clases","3. Solo funciones"], "r": "1"},
    {"p": "¿Qué comparten las clases?", "o": ["1. El mismo método","2. Variables iguales","3. Nada"], "r": "1"},
    {"p": "¿Qué cambia en el polimorfismo?", "o": ["1. El comportamiento","2. El nombre","3. Nada"], "r": "1"},
    {"p": "¿Qué es sobrescribir un método?", "o": ["1. Cambiar su comportamiento","2. Borrarlo","3. Duplicarlo"], "r": "1"},
    {"p": "¿El nombre del método cambia?", "o": ["1. Sí","2. No","3. A veces"], "r": "2"},
    {"p": "¿Dónde se usa el polimorfismo?", "o": ["1. Programación orientada a objetos","2. Excel","3. HTML"], "r": "1"},
    {"p": "¿Qué tienen las clases hijas?", "o": ["1. Métodos redefinidos","2. Nada","3. Solo variables"], "r": "1"},
    {"p": "¿El comportamiento es igual?", "o": ["1. Sí","2. No","3. Siempre"], "r": "2"},
    {"p": "¿Para qué sirve el polimorfismo?", "o": ["1. Flexibilidad en el código","2. Hacerlo más largo","3. Eliminar funciones"], "r": "1"}
]

random.shuffle(preguntas)

# =========================
# CLASE BASE
# =========================
class Personaje:
    def accion(self, pregunta):
        pass

# =========================
# POLIMORFISMO
# =========================
class Guerrero(Personaje):
    def accion(self, pregunta):
        print("\nTurno: Guerrero")
        return responder(pregunta, "Ataque con espada")

class Mago(Personaje):
    def accion(self, pregunta):
        print("\nTurno: Mago")
        return responder(pregunta, "Hechizo")

class Arquero(Personaje):
    def accion(self, pregunta):
        print("\nTurno: Arquero")
        return responder(pregunta, "Disparo")

# =========================
# FUNCIÓN RESPUESTA
# =========================
def responder(pregunta, estilo):
    print("\nPregunta:")
    print(pregunta["p"])

    for op in pregunta["o"]:
        print(op)

    r = input("Respuesta: ")

    if r == pregunta["r"]:
        print("Correcto -", estilo)
        return True
    else:
        print("Incorrecto")
        return False

# =========================
# JUEGO
# =========================
def elegir_personaje():
    print("Seleccione su personaje:")
    print("1. Guerrero")
    print("2. Mago")
    print("3. Arquero")

    op = input("Opción: ")

    if op == "1":
        return Guerrero()
    elif op == "2":
        return Mago()
    else:
        return Arquero()

def jugar():
    jugador = elegir_personaje()

    racha = 0
    fallos = 0

    print("\nObjetivo: responder 5 preguntas correctas consecutivas\n")

    for i, pregunta in enumerate(preguntas, start=1):
        print("\n----------------------------------")
        print(f"Pregunta {i} de 10")
        print(f"Racha actual: {racha} | Fallos: {fallos}")

        correcto = jugador.accion(pregunta)  # POLIMORFISMO

        if correcto:
            racha += 1
        else:
            racha = 0
            fallos += 1

        if racha == 5:
            print("\nResultado: GANASTE")
            return

        if fallos == 3:
            print("\nResultado: PERDISTE")
            return

    print("\nJuego terminado")

# =========================
# INICIO
# =========================
jugar()
```
