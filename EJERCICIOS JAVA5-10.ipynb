```python

# 5. Cuadratica


a = float(input("Ingrese el valor de a: "))
b = float(input("Ingrese el valor de b: "))
c = float(input("Ingrese el valor de c: "))

d = b * b - 4 * a * c

if d > 0:
    print("La ecuación tiene dos soluciones reales")
elif d == 0:
    print("La ecuación tiene una solución real")
else:
    print("La ecuación no tiene soluciones reales")



# 6. Par o Impar


n = int(input("Ingrese un número entero: "))

if n % 2 == 0:
    print("El número es par")
else:
    print("El número es impar")



# 7. Año Bisiesto


anio = int(input("Ingrese un año: "))

if (anio % 4 == 0 and anio % 100 != 0) or anio % 400 == 0:
    print("El año es bisiesto")
else:
    print("El año no es bisiesto")


# 8. Prioridad en Fila


nombre = input("Ingrese su nombre: ")
edad = int(input("Ingrese su edad: "))

if edad >= 70:
    print(nombre + " tiene prioridad en la fila")
else:
    print(nombre + " no tiene prioridad en la fila")



# 9. Mayor de Tres Numeros


a = int(input("Ingrese el primer número: "))
b = int(input("Ingrese el segundo número: "))
c = int(input("Ingrese el tercer número: "))

mayor = a

if b > mayor:
    mayor = b
if c > mayor:
    mayor = c

print("El número mayor es:", mayor)


# 10. Tipos de Triangulo


a = int(input("Ingrese el primer lado: "))
b = int(input("Ingrese el segundo lado: "))
c = int(input("Ingrese el tercer lado: "))

if a == b and b == c:
    print("El triángulo es equilátero")
elif a == b or a == c or b == c:
    print("El triángulo es isósceles")
else:
    print("El triángulo es escaleno")



# 11. IMC

nombre = input("Ingrese su nombre: ")
peso = float(input("Ingrese su peso en kg: "))
altura = float(input("Ingrese su altura en metros: "))

imc = peso / (altura * altura)

if imc < 18.5:
    print(nombre + ", su IMC indica bajo peso")
elif imc < 25:
    print(nombre + ", su IMC es normal")
elif imc < 30:
    print(nombre + ", su IMC indica sobrepeso")
else:
    print(nombre + ", su IMC indica obesidad")



# 12. Calculadora Simple


a = float(input("Ingrese el primer número: "))
b = float(input("Ingrese el segundo número: "))
op = input("Ingrese la operación (+ - * /): ")

if op == "+":
    print("Resultado:", a + b)
elif op == "-":
    print("Resultado:", a - b)
elif op == "*":
    print("Resultado:", a * b)
elif op == "/":
    print("Resultado:", a / b)



# 13. Nota en Letra


nota = int(input("Ingrese la nota final: "))

if nota >= 90:
    print("Calificación: A")
elif nota >= 80:
    print("Calificación: B")
elif nota >= 70:
    print("Calificación: C")
elif nota >= 60:
    print("Calificación: D")
else:
    print("Calificación: F")



# 14. Dias de la Semana


d = int(input("Ingrese un número del 1 al 7: "))

if d == 1:
    print("Lunes")
elif d == 2:
    print("Martes")
elif d == 3:
    print("Miércoles")
elif d == 4:
    print("Jueves")
elif d == 5:
    print("Viernes")
elif d == 6:
    print("Sábado")
elif d == 7:
    print("Domingo")



# 15. Formar Triangulo


a = int(input("Ingrese el primer lado: "))
b = int(input("Ingrese el segundo lado: "))
c = int(input("Ingrese el tercer lado: "))

if a + b > c and a + c > b and b + c > a:
    if a == b and b == c:
        print("El triángulo es equilátero")
    elif a == b or a == c or b == c:
        print("El triángulo es isósceles")
    else:
        print("El triángulo es escaleno")
else:
    print("Los valores no forman un triángulo")
```
