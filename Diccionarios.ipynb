````python
galleta=["dinosaurio","flores","carro"]
precio=[5000,6000,7000]
cant=[5,5,6]
````

````python
def menu():
  for galletas in range(3):
    print(f"galleta={galleta[galletas]}, precio={precio[galletas]}, cantidad={cant[galletas]}")
````

````python
menu()
````
