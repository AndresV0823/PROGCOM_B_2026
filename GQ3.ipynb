```python
#Andres Valencia
from typing import List, Dict

class Gasto:
    def __init__(self, nombre: str, categoria: str, monto: float):
        self.nombre = nombre
        self.categoria = categoria
        self.monto = monto

    def __repr__(self):
        return f"{self.nombre} ({self.categoria}) = ${self.monto:.2f}"

class PeriodoFinanciero:
    def __init__(self, tipo_periodo: str, presupuesto: float):
        self.tipo_periodo = tipo_periodo  # diario, semanal, quincenal, mensual
        self.presupuesto = presupuesto
        self.gastos: List[Gasto] = []
        self.ingresos = 0.0

    def agregar_gasto(self, gasto: Gasto):
        self.gastos.append(gasto)

    def registrar_ingresos(self, monto: float):
        self.ingresos += monto

    def calcular_balance(self):
        total_gastos = sum(g.monto for g in self.gastos)
        balance = self.ingresos - total_gastos
        presupuesto_restante = self.presupuesto - total_gastos
        return balance, presupuesto_restante

    def generar_estado_financiero(self):
        balance, presupuesto_restante = self.calcular_balance()
        print("\n--- ESTADO FINANCIERO ---")
        print(f"Período: {self.tipo_periodo.capitalize()}")
        print(f"Presupuesto: ${self.presupuesto:.2f}")
        print(f"Ingresos: ${self.ingresos:.2f}")
        print(f"Gastos totales: ${sum(g.monto for g in self.gastos):.2f}")
        print(f"Balance neto: ${balance:.2f}")
        print(f"Presupuesto restante: ${presupuesto_restante:.2f}\n")

        # Análisis financiero simple
        if balance > 0 and presupuesto_restante >= 0:
            print("Estado: Positivo / Sano. Buena liquidez y dentro del presupuesto.")
        elif balance > 0 and presupuesto_restante < 0:
            print("Estado: Positivo pero sobrepasó presupuesto. Vigilar gastos.")
        elif balance < 0 and presupuesto_restante >= 0:
            print("Estado: Pérdida neta, pero aún dentro del presupuesto. Revisar ingresos.")
        else:
            print("Estado: Pérdida y sobrepasó presupuesto. Acción inmediata requerida.")

def main():
    print("=== OPTIMIZADOR DE INGRESOS Y GASTOS ===\n")
    print("Seleccione el período de análisis:")
    print("1. Diario")
    print("2. Semanal")
    print("3. Quincenal")
    print("4. Mensual")
    opcion = input("Ingrese el número del período: ")

    periodos = {"1": "diario", "2": "semanal", "3": "quincenal", "4": "mensual"}
    tipo_periodo = periodos.get(opcion, "mensual")

    while True:
        try:
            presupuesto = float(input(f"Ingrese el presupuesto disponible para este período ({tipo_periodo}): "))
            if presupuesto < 0:
                raise ValueError
            break
        except ValueError:
            print("Ingrese un número válido mayor o igual a cero.")

    periodo = PeriodoFinanciero(tipo_periodo, presupuesto)

    print("\n--- REGISTRO DE GASTOS ---")
    while True:
        nombre = input("Nombre del gasto (o 'fin' para terminar): ")
        if nombre.lower() == 'fin':
            break
        categoria = input("Categoría del gasto: ")
        while True:
            try:
                monto = float(input("Monto del gasto: "))
                if monto < 0:
                    raise ValueError
                break
            except ValueError:
                print("Ingrese un monto válido mayor o igual a cero.")
        periodo.agregar_gasto(Gasto(nombre, categoria, monto))
        total_gastos = sum(g.monto for g in periodo.gastos)
        if total_gastos > periodo.presupuesto:
            print(f"¡Atención! Se ha sobrepasado el presupuesto en ${total_gastos - periodo.presupuesto:.2f}")

    while True:
        try:
            ingresos = float(input("\nIngrese los ingresos totales de este período: "))
            if ingresos < 0:
                raise ValueError
            periodo.registrar_ingresos(ingresos)
            break
        except ValueError:
            print("Ingrese un número válido mayor o igual a cero.")

    # Generar reporte financiero
    periodo.generar_estado_financiero()

    # Detalle de gastos por categoría
    categorias: Dict[str, float] = {}
    for gasto in periodo.gastos:
        categorias[gasto.categoria] = categorias.get(gasto.categoria, 0) + gasto.monto
    print("\n--- DISTRIBUCIÓN DE GASTOS POR CATEGORÍA ---")
    for cat, monto in categorias.items():
        print(f"{cat}: ${monto:.2f}")

if __name__ == "__main__":
    main()
```
