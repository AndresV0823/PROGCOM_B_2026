```python
# Sistema de producción flexible con días mensuales

class Produccion:
    def __init__(self, meta, tipo_meta):
        self.meta = meta
        self.tipo_meta = tipo_meta
        self.registros = []

    def agregar_registro(self, periodo, cantidad):
        self.registros.append((periodo, cantidad))

    def total_producido(self):
        return sum(cantidad for _, cantidad in self.registros)

    def calcular_eficiencia(self):
        total = self.total_producido()
        return (total / self.meta) * 100

    def mostrar_resumen(self):
        print("\n--- RESUMEN DE PRODUCCIÓN ---")

        for periodo, cantidad in self.registros:
            print(f"{periodo} -> {cantidad} unidades")

        total = self.total_producido()
        eficiencia = self.calcular_eficiencia()

        print(f"\nTotal producido: {total} unidades")
        print(f"Meta {self.tipo_meta}: {self.meta} unidades")
        print(f"Eficiencia: {round(eficiencia, 2)}%")

        if total >= self.meta:
            print("Estado: Meta cumplida")
        else:
            print("Estado: Meta NO cumplida")


def main():
    print("SISTEMA DE PRODUCCIÓN FLEXIBLE\n")

    # Elegir tipo de registro
    while True:
        tipo_registro = input("¿Registrar por (1) Días o (2) Semanas?: ")
        if tipo_registro in ["1", "2"]:
            break
        print("Opción inválida.")

    # Lógica de selección de meta
    if tipo_registro == "1":
        while True:
            tipo_meta = input("¿Meta (1) Semanal o (2) Mensual?: ")
            if tipo_meta == "1":
                tipo_meta_texto = "semanal"
                break
            elif tipo_meta == "2":
                tipo_meta_texto = "mensual"
                break
            print("Opción inválida.")
    else:
        tipo_meta_texto = "mensual"
        print("Meta configurada automáticamente como mensual.\n")

    meta = int(input(f"Ingrese la meta {tipo_meta_texto}: "))
    sistema = Produccion(meta, tipo_meta_texto)

    print("\nIngrese los datos:\n")

    if tipo_registro == "1":
        # Determinar número de días a registrar
        if tipo_meta_texto == "semanal":
            total_periodos = 7
        else:  # mensual
            total_periodos = 30

        for i in range(1, total_periodos + 1):
            while True:
                try:
                    cantidad = int(input(f"Día {i}: "))
                    if cantidad < 0:
                        print("Valor inválido.")
                        continue
                    break
                except ValueError:
                    print("Entrada inválida.")
            sistema.agregar_registro(f"Día {i}", cantidad)

    else:
        # Registro por semanas (4 semanas)
        for i in range(1, 5):
            while True:
                try:
                    cantidad = int(input(f"Semana {i}: "))
                    if cantidad < 0:
                        print("Valor inválido.")
                        continue
                    break
                except ValueError:
                    print("Entrada inválida.")
            sistema.agregar_registro(f"Semana {i}", cantidad)

    sistema.mostrar_resumen()


if __name__ == "__main__":
    main()
```

```python
import tkinter as tk
from tkinter import filedialog, messagebox
import json

root = tk.Tk()
root.title("Sistema de Producción - PRO")
root.geometry("900x650")

registros = []

# =================== FUNCIONES ===================

def actualizar_lista():
    lista.delete(0, tk.END)
    for r in registros:
        lista.insert(tk.END, f"{r['periodo']} → {r['cantidad']} unidades")


def agregar():
    try:
        p = periodo.get()
        c = int(cantidad.get())

        if p == "":
            return

        registros.append({"periodo": p, "cantidad": c})

        periodo.delete(0, tk.END)
        cantidad.delete(0, tk.END)

        actualizar_lista()
        guardar_auto()

    except:
        messagebox.showerror("Error", "Datos inválidos")


def calcular():
    try:
        meta = int(meta_entry.get())
        total = sum(r["cantidad"] for r in registros)

        resultado.delete(1.0, tk.END)

        resultado.insert(tk.END, "=== RESUMEN ===\n\n")

        for r in registros:
            resultado.insert(tk.END, f"{r['periodo']} → {r['cantidad']}\n")

        resultado.insert(tk.END, f"\nTOTAL: {total}\nMETA: {meta}\n")

        if total >= meta:
            resultado.insert(tk.END, "ESTADO: META CUMPLIDA")
        else:
            resultado.insert(tk.END, "ESTADO: META NO CUMPLIDA")

    except:
        messagebox.showerror("Error", "Meta inválida")


def guardar():
    archivo = filedialog.asksaveasfilename(defaultextension=".json")
    if archivo:
        with open(archivo, "w") as f:
            json.dump(registros, f)


def abrir():
    archivo = filedialog.askopenfilename()
    if archivo:
        with open(archivo, "r") as f:
            datos = json.load(f)

        registros.clear()
        registros.extend(datos)

        actualizar_lista()
        calcular()


def eliminar():
    try:
        i = lista.curselection()[0]
        registros.pop(i)
        actualizar_lista()
        guardar_auto()
    except:
        messagebox.showerror("Error", "Selecciona un elemento")


def guardar_auto():
    with open("autosave.json", "w") as f:
        json.dump(registros, f)


def cargar_auto():
    try:
        with open("autosave.json", "r") as f:
            datos = json.load(f)
        registros.extend(datos)
        actualizar_lista()
    except:
        pass


# =================== UI ===================

# TOP PANEL
top = tk.Frame(root)
top.pack(pady=10)

tk.Label(top, text="Periodo").grid(row=0, column=0)
periodo = tk.Entry(top, width=15)
periodo.grid(row=1, column=0)

tk.Label(top, text="Cantidad").grid(row=0, column=1)
cantidad = tk.Entry(top, width=15)
cantidad.grid(row=1, column=1)

tk.Label(top, text="Meta").grid(row=0, column=2)
meta_entry = tk.Entry(top, width=15)
meta_entry.grid(row=1, column=2)

# BOTONES
botones = tk.Frame(root)
botones.pack(pady=10)

tk.Button(botones, text="Agregar", width=12, command=agregar).grid(row=0, column=0, padx=5)
tk.Button(botones, text="Calcular", width=12, command=calcular).grid(row=0, column=1, padx=5)
tk.Button(botones, text="Guardar", width=12, command=guardar).grid(row=0, column=2, padx=5)
tk.Button(botones, text="Abrir", width=12, command=abrir).grid(row=0, column=3, padx=5)
tk.Button(botones, text="Eliminar", width=12, command=eliminar).grid(row=0, column=4, padx=5)

# LISTA
lista = tk.Listbox(root, width=80, height=12)
lista.pack(pady=10)

# RESULTADO
resultado = tk.Text(root, height=15, width=90)
resultado.pack(pady=10)

# INICIO
cargar_auto()
root.mainloop()
```
