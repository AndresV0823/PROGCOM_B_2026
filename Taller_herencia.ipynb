```python
class Vehiculo:
    def __init__(self, marca):
        self.marca = marca

class Carro(Vehiculo):
    def __init__(self, marca, modelo):
        super().__init__(marca)
        self.modelo = modelo

c = Carro("Toyota", "Corolla")
print(c.marca)
print(c.modelo)
```

```python
class Persona:
    def __init__(self, nombre):
        self.nombre = nombre

class Estudiante(Persona):
    def __init__(self, nombre, curso):
        super().__init__(nombre)
        self.curso = curso

e = Estudiante("Juan", "11B")
print(e.nombre)
print(e.curso)
```

```python
class Animal:
    def __init__(self, nombre):
        self.nombre = nombre

class Perro(Animal):
    def __init__(self, nombre, raza):
        super().__init__(nombre)
        self.raza = raza

p = Perro("Rex", "Labrador")
print(p.nombre)
print(p.raza)
```

```python
class Ropa:
    def __init__(self, color):
        self.color = color

class Camisa(Ropa):
    def __init__(self, color, talla):
        super().__init__(color)
        self.talla = talla

cam = Camisa("Azul", "M")
print(cam.color)
print(cam.talla)
```

```python
class Dispositivo:
    def __init__(self, marca):
        self.marca = marca

class Celular(Dispositivo):
    def __init__(self, marca, memoria):
        super().__init__(marca)
        self.memoria = memoria

cel = Celular("Samsung", "128GB")
print(cel.marca)
print(cel.memoria)
```

```python
class Persona:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad
    def saludar(self):
        print("Hola, soy", self.nombre)

class Profesor(Persona):
    def __init__(self, nombre, edad, materia):
        super().__init__(nombre, edad)
        self.materia = materia

prof = Profesor("Carlos", 35, "Matemáticas")
print(prof.nombre)
print(prof.edad)
print(prof.materia)
```

```python
class Cuenta:
    def __init__(self, titular, saldo):
        self.titular = titular
        self.saldo = saldo
    def mostrar_saldo(self):
        print("Saldo:", self.saldo)

class CuentaCorriente(Cuenta):
    def __init__(self, titular, saldo, numero_cuenta):
        super().__init__(titular, saldo)
        self.numero_cuenta = numero_cuenta

cc = CuentaCorriente("Ana", 500000, "001-2345")
print(cc.titular)
print(cc.numero_cuenta)
print(cc.saldo)
```

```python
class Empleado:
    def __init__(self, nombre, salario):
        self.nombre = nombre
        self.salario = salario
    def trabajar(self):
        print(self.nombre, "está trabajando")

class Disenador(Empleado):
    def __init__(self, nombre, salario, programa):
        super().__init__(nombre, salario)
        self.programa = programa

d = Disenador("Laura", 2000000, "Photoshop")
print(d.nombre)
print(d.salario)
print(d.programa)
```

```python
class Figura:
    def mostrar_figura(self):
        print("Soy una figura geométrica")

class Triangulo(Figura):
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura
    def calcular_area(self):
        return (self.base * self.altura) / 2

t = Triangulo(5, 8)
t.mostrar_figura()
print("Área:", t.calcular_area())
```

```python
class Deportista:
    def __init__(self, nombre, deporte):
        self.nombre = nombre
        self.deporte = deporte
    def entrenar(self):
        print(self.nombre, "está entrenando")

class Nadador(Deportista):
    def __init__(self, nombre, deporte, estilo):
        super().__init__(nombre, deporte)
        self.estilo = estilo

n = Nadador("Miguel", "Natación", "Mariposa")
print(n.nombre)
print(n.deporte)
print(n.estilo)
```
